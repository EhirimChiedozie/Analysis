```python
from matplotlib import pyplot as pyplot

plt.title('''Bar Chart showing the number of goals
          scored this season by some Arsenal players''')
plt.xlabel('Players')
plt.ylabel('Number of Goals')
plt.bar(df['Names'], df['Goals'],)
plt.show
```

```python
plt.style.use('fivethirtyeight')
wedgeprops = {'edgecolor': 'black'}
plt.title('''Pie Chart showing the number of assists 
          this season by some Arsenal players''')
labels = [i for i in df['Names']]
plt.pie(df['Assists'], labels=labels, wedgeprops=wedgeprops)
plt.tight_layout()
plt.show()
```

```python

w, x = 0.4, numpy.arange(len(df))
plt.style.use('fivethirtyeight')
plt.xlabel('Names of Players')
plt.ylabel('Number of Goals and Assists')
plt.title('''Multiple Bar Chart showing the number of goals 
          and assists this season by some Arsenal players''')
plt.bar(x - w/2, df['Goals'], w, label='Goals')
plt.bar(x + w/2, df['Assists'], w, label='Assists')
plt.xticks(x, df['Names'])
plt.legend()
plt.show()
```
