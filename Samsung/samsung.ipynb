```python
import pandas
from collections import Counter
from matplotlib import pyplot as plt
```

```python
df = pandas.read_csv('Expanded_Dataset.csv')
df.head()
```

```python
def all_phones_by_year(year):
    year_df = df.loc[df['Year'] == year]
    return year_df['Units Sold'].sum()
```

```python
distinct_year = [Counter(df['Year'])[0]]
```

```python
def total_revenue_by_year(year):
    year_df = df.loc[df['Year'] == year]
    return year_df['Revenue ($)'].sum()
```

```python
total_revenue_by_year(2019)
```

In 2019 alone, the company recorded a total of 5,169,832(Five million, one hundred and sixty nine thousand, eight hundred and thirty two) Samsung phones and generated a total revenue of $5,607,306,561(Five billion six and seven million, three hundred and six thousand, five hundred and sixty one dollars).

Now, lets check how many of these phones have 5G capability

```python
len(Counter(df['Product Model']))
```

```python
cap_5g = df.loc[df['5G Capability'] == 'Yes']
length5GYes = len(cap_5g['Product Model'].value_counts())
length5GYes
```

```python
no5g_cap = df.loc[df['5G Capability'] == 'No']
length5GNo = len(no5g_cap['Product Model'].value_counts())
length5GNo
```

There are 15 different Samsung models in  the dataset. 10 of them support 5G capability and 5 of them do not support 5G capability.

Now, let us check the region with the most 5G samsung phones

```python
region5g = Counter(cap_5g['Region']).most_common()
region5g
```

```python
countries = [region5g[i][0] for i in range(len(region5g))]
country_count = [region5g[i][1] for i in range(len(region5g))]
plt.pie(country_count, labels=countries)
plt.show()
```
